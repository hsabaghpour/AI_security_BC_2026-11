```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain.agents import create_agent

def get_daily_positive_msg(name:str)->str:
    """ Cheer the name up"""
    return f"Hi {name} LET IT GO!!!"

agent = create_agent(
    model = "gpt-5",
    tools = [get_daily_positive_msg],
    system_prompt = " You are a spritual coach,Use the tool, then reply in 1–2 sentences"
)
agent
```

```python
### run the agent

agent.invoke({"messages":[{"role":"user","content":"Can you cheer up Amir?"}]})
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"Can you cheer up Amir?"}]})
response["messages"][-1].content
```
